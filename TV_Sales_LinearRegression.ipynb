```python
import numpy as np
import pandas as pd
from sklearn import linear_model
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import r2_score
```

```python
data=pd.read_csv(r'C:\Users\hp\Downloads\tvmarketing-1-1_1729437291501 (1).csv')
```

```python
data
```

```python
data.describe()
```

We can see from the above table the data is very well destributed so balanced data

```python
x=data[['TV']]
```

```python
y=data[['Sales']]
```

```python
model=linear_model.LinearRegression()
```

```python
model.fit(x,y)
```

Now lets see how exactly the regression line fits 

```python
sns.regplot(x="TV",y="Sales",data=data)
```

```python
#note when we pass x values to the model predict it will predict the sales value
predictions=model.predict(x)
data['pred']=np.round(predictions,1)
```

```python
data
```

```python
#as we can see the actuals and the prected values side by side here
r2_score(data[['Sales']],data[['pred']])
```

CONCLUSION 1: -

We got r2_score as 61% which is slightly less than what we expect in a good linear regression model but as we can see the data has only 1 feture and a comparitivly less input data so the prediction score is very practical and realistic for this model

```python
# TO IMPROVE THIS WE CAN USE POLINOMIAL REGRESSION
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly=PolynomialFeatures(degree=2, include_bias=False)

x_poly=poly.fit_transform(x)

poly_model=LinearRegression()

poly_model.fit(x_poly,y)
```

```python
y_pred_new=poly_model.predict(x_poly)
r2_score_poly=r2_score(y,y_pred_new)
```
